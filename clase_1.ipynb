### Ejercicio 6 (Propuesto - Ejercicio Rápido): Mapeo de Rangos de Satisfacción Textual
Se agrega una columna categórica de texto: `rango_satisfaccion` con valores 'Bajo', 'Medio', 'Alto'. Crea una función que realice la codificación jerárquica respetando el orden: Bajo -> 1, Medio -> 2, Alto -> 3.

```python
def codificar_satisfaccion_texto(df_input: pd.DataFrame) -> pd.DataFrame:
    df_temp = df_input.copy()
    # Generamos la columna de texto de ejemplo basándonos en los valores de satisfacción actuales
    bins = [0, 2.5, 4.0, 6.0]
    labels = ["Bajo", "Medio", "Alto"]
    df_temp["rango_satisfaccion"] = pd.cut(df_temp["satisfaccion"].fillna(3.0), bins=bins, labels=labels).astype(str)
    dicci={"Bajo":1,"Medio":2,"Alto":3}
    df_temp=df_temp["rango_satisfaccion"].map(dicci)
    return df_temp


# --- PRUEBA UNITARIA ---
try:
    df_orig = pd.read_csv("datos_clientes.csv")
    df_res = codificar_satisfaccion_texto(df_orig)
    val_alto = df_res[df_res["rango_satisfaccion"] == "Alto"]["rango_satisfaccion_cod"].iloc[0]
    val_bajo = df_res[df_res["rango_satisfaccion"] == "Bajo"]["rango_satisfaccion_cod"].iloc[0]
    assert val_alto == 3, f"Error: Alto debería ser 3, es {val_alto}"
    assert val_bajo == 1, f"Error: Bajo debería ser 1, es {val_bajo}"
    print("¡Ejercicio 6 superado con éxito!")
except Exception as e:
    print(f"Prueba fallida: {type(e).__name__} -> {e}")
```

### Ejercicio 7 (Resuelto): Estadísticas de Dispersión Robustas
Calculamos y comparamos la desviación estándar clásica frente al Rango Intercuartílico (IQR) y la Desviación Absoluta de la Mediana (MAD) para el ingreso.

```python
from scipy import stats

df = pd.read_csv("datos_clientes.csv")
df_limpio = df.dropna(subset=["ingreso"])

# Medida clásica
desviacion_estandar = df_limpio["ingreso"].std()

# Medidas robustas
q1 = df_limpio["ingreso"].quantile(0.25)
q3 = df_limpio["ingreso"].quantile(0.75)
iqr = q3 - q1
mad = stats.median_abs_deviation(df_limpio["ingreso"])

print(f"Desviación Estándar Clásica: {desviacion_estandar:.2f} (muy inflada por el outlier)")
print(f"Rango Intercuartílico (IQR): {iqr:.2f} (robusto)")
print(f"MAD (Desviación Absoluta de la Mediana): {mad:.2f} (robusta)")
```

### Ejercicio 8 (Resuelto): Detección de Outliers y Visualización Avanzada (Boxplot y Dispersión)

Implementamos la regla de Tukey para detectar outliers cuantitativos. Para analizarlos de forma visual:
1. Trazamos un diagrama de caja (**boxplot**) para ver cómo se sitúan fuera de los bigotes.
2. Trazamos un gráfico de **dispersión** (scatterplot de `edad` vs. `ingreso`) resaltando los outliers detectados en color rojo con un marcador diferente para entender su posición bidimensional.

```python
df_out = pd.read_csv("datos_clientes.csv").dropna(subset=["ingreso"])

# 1. Definir límites de Tukey
q1 = df_out["ingreso"].quantile(0.25)
q3 = df_out["ingreso"].quantile(0.75)
iqr = q3 - q1
limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

df_out["es_outlier"] = (df_out["ingreso"] < limite_inferior) | (df_out["ingreso"] > limite_superior)
print(f"Límites de Tukey: Inferior = {limite_inferior:.2f}, Superior = {limite_superior:.2f}")
print(f"Outliers detectados:\n", df_out[df_out["es_outlier"] == True][["cliente_id", "ingreso"]])

# 2. Gráfico 1: Boxplot (excluyendo el outlier extremo de 999k para escala visual)
plt.figure(figsize=(8, 2.5))
df_box = df_out[df_out["ingreso"] < 120000]
sns.boxplot(x=df_box["ingreso"], color="orange")
plt.title("Boxplot de Ingresos (excluyendo outlier > 900k)")
plt.show()

# 3. Gráfico 2: Dispersión de Edad vs. Ingreso (resaltando outliers en rojo)
plt.figure(figsize=(8, 4.5))
sns.scatterplot(
    data=df_out,
    x="edad",
    y="ingreso",
    hue="es_outlier",
    palette={False: "#3880F8", True: "#E02424"},
    style="es_outlier",
    markers={False: "o", True: "X"},
    s=100
)
plt.title("Dispersión Edad vs. Ingreso (Outliers en Rojo/X)")
plt.xlabel("Edad")
plt.ylabel("Ingreso")
plt.legend(title="¿Es Outlier?")
plt.show()
```

### Ejercicio 9 (Propuesto - Ejercicio Rápido): Filtrado de Registros Outliers usando Tukey
Implementa una función para calcular los límites de Tukey sobre el DataFrame e identificar y excluir todos los outliers de la columna `ingreso`, retornando únicamente el conjunto de registros inliers.

```python
def filtrar_outliers_ingreso(df_input: pd.DataFrame) -> pd.DataFrame:
    df_temp = df_input.copy().dropna(subset=["ingreso"])
    # TODO: Calcula Q1, Q3 y el IQR para la columna 'ingreso'.
    # Define los límites inferior y superior con un factor de 1.5.
    # Filtra el DataFrame eliminando los outliers (mantén solo los inliers).
    # Retorna el DataFrame filtrado.
    q1 = df_temp["ingreso"].quantile(0.25)
    q3 = df_temp["ingreso"].quantile(0.75)
    iqr = q3 - q1
    limite_inferior = q1 - 1.5 * iqr
    limite_superior = q3 + 1.5 * iqr

    return df_temp[(df_temp["ingreso"] > limite_inferior) & (df_temp["ingreso"] < limite_superior)]

# --- PRUEBA UNITARIA ---
try:
    df_orig = pd.read_csv("datos_clientes.csv")
    df_inliers = filtrar_outliers_ingreso(df_orig)
    print(f"Registros originales con ingreso: {df_orig['ingreso'].notna().sum()}")
    print(f"Registros inliers resultantes: {len(df_inliers)}")
    
    assert 999999.0 not in df_inliers["ingreso"].values, "Error: Outlier extremo no fue eliminado."
    assert -5000.0 not in df_inliers["ingreso"].values, "Error: Outlier negativo no fue eliminado."
    assert len(df_inliers) == 90, f"Error: Se esperaban 90 inliers, se obtuvieron {len(df_inliers)}"
    print("¡Ejercicio 9 superado con éxito!")
except Exception as e:
    print(f"Prueba fallida: {type(e).__name__} -> {e}")
```

## 5. Tema 4: Análisis de Distribución y Pruebas de Normalidad

Muchos modelos de ciencia de datos asumen una distribución normal. Estudiamos pruebas de hipótesis y transformaciones para normalización.

### Ejercicio 10 (Resuelto): Prueba Formal de Shapiro-Wilk y Visualización Q-Q Plot
Ejecutamos la prueba de normalidad de Shapiro-Wilk sobre datos normales y visualizamos su ajuste mediante un gráfico de cuantiles teóricos (**Gráfico Q-Q**).

```python
# 1. Generamos 50 valores con distribución normal
np.random.seed(42)
datos_normales = np.random.normal(loc=100, scale=15, size=50)

# 2. Ejecutar prueba Shapiro-Wilk
estadistico, p_valor = stats.shapiro(datos_normales)
print(f"Estadístico Shapiro: {estadistico:.4f}, p-valor: {p_valor:.4f}")
if p_valor >= 0.05:
    print("H0 no se rechaza: Los datos tienen una distribución normal (p >= 0.05)")
else:
    print("H0 rechazada: Los datos NO siguen una distribución normal (p < 0.05)")

# 3. Gráfico Q-Q de cuantiles para evaluar ajuste visual
plt.figure(figsize=(6, 4))
stats.probplot(datos_normales, dist="norm", plot=plt)
plt.title("Gráfico Cuantil-Cuantil (Q-Q Plot)")
plt.xlabel("Cuantiles Teóricos Normales")
plt.ylabel("Cuantiles Observados")
plt.show()
```

### Ejercicio 11 (Resuelto): Normalización por Yeo-Johnson y Visualización de Distribución

Aplicamos la transformación matemática **Yeo-Johnson** de SciPy para corregir la asimetría de la columna `ingreso`. Visualizamos el cambio en la distribución mostrando el antes y después en dos subgráficos alineados.

### Ejercicio 12 (Propuesto - Ejercicio Rápido): Evaluar Normatividad de una Variable
Completa la función `evaluar_normalidad` para que ejecute la prueba de Shapiro-Wilk y retorne `True` si los datos siguen una distribución normal (p-valor >= 0.05) y `False` de lo contrario.

```python
def evaluar_normalidad(datos: np.ndarray) -> bool:
    estadistico, p_valor = stats.shapiro(datos)
    return p_valor >= 0.05

# --- PRUEBA UNITARIA ---
try:
    # Caso 1: Normal
    np.random.seed(42)
    datos_norm = np.random.normal(loc=50, scale=5, size=100)
    res_norm = evaluar_normalidad(datos_norm)
    
    # Caso 2: Uniforme (no normal)
    datos_unif = np.random.uniform(low=0, high=100, size=100)
    res_unif = evaluar_normalidad(datos_unif)
    
    assert res_norm == True, "Error: Los datos normales deberían retornar True."
    assert res_unif == False, "Error: Los datos uniformes deberían retornar False."
    print("¡Ejercicio 12 superado con éxito!")
except Exception as e:
    print(f"Prueba fallida: {type(e).__name__} -> {e}")
```
