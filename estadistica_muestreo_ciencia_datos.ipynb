# 🏛️ 1. Población vs. Muestra: Conceptos, Diferencias y Notación

### 📖 Conceptos Fundamentales:
* **Población ($N$)**: Es el conjunto completo de todos los elementos, individuos, eventos o mediciones de interés para una investigación.
  * Los valores numéricos que describen a la población se denominan **Parámetros** (ej. Media poblacional $\mu$, Desviación estándar poblacional $\sigma$, Proporción poblacional $P$).
* **Muestra ($n$)**: Es un subconjunto representativo extraído de la población para realizar inferencias sobre ella.
  * Los valores numéricos calculados a partir de una muestra se denominan **Estadísticos** o **Estimadores** (ej. Media muestral $\bar{x}$, Desviación estándar muestral $s$, Proporción muestral $p$).

| Característica | Población (Universo) | Muestra (Subconjunto) |
| :--- | :--- | :--- |
| **Definición** | Colección total de elementos | Subconjunto representativo |
| **Tamaño** | Simbolizado por $N$ (usualmente grande/infinito) | Simbolizado por $n$ ($n < N$) |
| **Métrica calculada** | **Parámetro** (Fijo, usualmente desconocido) | **Estadístico** (Variable aleatoria) |
| **Media** | $\mu = \frac{\sum X}{N}$ | $\bar{x} = \frac{\sum x}{n}$ |
| **Varianza** | $\sigma^2 = \frac{\sum (X - \mu)^2}{N}$ | $s^2 = \frac{\sum (x - \bar{x})^2}{n - 1}$ *(corrección de Bessel)* |
| **Costo/Tiempo** | Alto / Frecuentemente inviable | Eficiente y controlable |

### 🎯 Storytelling:
Un buen estadístico debe ser **insesgado** ($E[\bar{x}] = \mu$) y **consistente** (a mayor $n$, $s \to \sigma$). Veamos 4 escenarios del mundo real con poblaciones de **$N = 1,000$ registros**.



````python
# ==============================================================================
# EJEMPLO 1.1: Salarios de Data Scientists en Tech (Distribución Log-Normal)
# ==============================================================================
# Población N=1000 salarios en USD con asimetría positiva típica
N = 1000
salarios_poblacion = np.random.lognormal(mean=11.2, sigma=0.45, size=N) # Media ~ $80k, colas largas
df_salarios = pd.DataFrame({'Salario_USD': salarios_poblacion})

# Extraemos una muestra aleatoria n=100
n = 100
df_muestra_salarios = df_salarios.sample(n=n, random_state=SEED)

# Parámetros vs Estadísticos
mu_salario = df_salarios['Salario_USD'].mean()
sigma_salario = df_salarios['Salario_USD'].std(ddof=0)
xbar_salario = df_muestra_salarios['Salario_USD'].mean()
s_salario = df_muestra_salarios['Salario_USD'].std(ddof=1)
error_estimacion = abs(xbar_salario - mu_salario)

print(f"--- EJEMPLO 1.1: SALARIOS TECH ---")
print(f"Población N={N}  | Media Real (μ): ${mu_salario:,.2f} | Desv. Est (σ): ${sigma_salario:,.2f}")
print(f"Muestra   n={n}   | Media Muestral (x̄): ${xbar_salario:,.2f} | Desv. Est (s): ${s_salario:,.2f}")
print(f"Error absoluto de estimación: ${error_estimacion:,.2f} ({(error_estimacion/mu_salario)*100:.2f}% de error)")

# Visualización con Storytelling
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))

# Histograma & KDE Comparativo
sns.histplot(df_salarios['Salario_USD'], kde=True, color=PRIMARY_COLOR, ax=ax1, stat='density', alpha=0.4, label=f'Población (N={N})')
sns.histplot(df_muestra_salarios['Salario_USD'], kde=True, color=SAMPLE_COLOR, ax=ax1, stat='density', alpha=0.5, label=f'Muestra (n={n})')
ax1.axvline(mu_salario, color='blue', linestyle='--', linewidth=2, label=f'μ = ${mu_salario:,.0f}')
ax1.axvline(xbar_salario, color='orange', linestyle='-', linewidth=2, label=f'x̄ = ${xbar_salario:,.0f}')
ax1.set_title("1.1 Distribución Salarial: Población vs Muestra")
ax1.set_xlabel("Salario Anual (USD)")
ax1.legend()

# Boxplot
sns.boxplot(data=[df_salarios['Salario_USD'], df_muestra_salarios['Salario_USD']], orient='h', palette=[PRIMARY_COLOR, SAMPLE_COLOR], ax=ax2)
ax2.set_yticklabels([f'Población (N={N})', f'Muestra (n={n})'])
ax2.set_title("Boxplot Comparativo de Dispersión y Outliers")
ax2.set_xlabel("Salario Anual (USD)")

plt.tight_layout()
plt.show()
````

# 🎲 2. Muestreo Aleatorio Simple (MAS / Simple Random Sampling)

### 📖 Concepto:
El **Muestreo Aleatorio Simple (MAS)** es la técnica básica donde **cada elemento de la población tiene exactamente la misma probabilidad ($1/N$)** de ser seleccionado. Además, cada subconjunto de tamaño $n$ tiene la misma probabilidad de ser la muestra elegida.

* **Con Reemplazo (MASCR)**: Un elemento seleccionado puede volver a elegirse. Los eventos son estadísticamente independientes.
* **Sin Reemplazo (MASSR)**: Un elemento seleccionado no puede volver a elegirse. Es el estándar en encuestas y análisis de datos.

### 📐 Fórmulas Clave:
* Probabilidad de selección de cada individuo: $P(i) = \frac{1}{N}$
* Varianza del estimador de la media: $Var(\bar{x}) = \frac{\sigma^2}{n} \left(\frac{N - n}{N - 1}\right)$ *(el segundo término es el Factor de Corrección por Población Finita - FCPF)*.

### 🌟 Ventajas y Desventajas:
* **Ventajas**: Fácil de implementar, no requiere información previa sobre subgrupos, genera estimaciones insesgadas.
* **Desventajas**: Si la población está dispersa geográficamente es costoso; puede subrepresentar minorías críticas si $n$ es pequeño.

A continuación, implementamos **4 ejemplos aplicados** con $N = 1,000$ registros cada uno.



````python
# ==============================================================================
# EJEMPLO 2.1: Encuesta de Satisfacción de Clientes CSAT (Escala 1 - 100)
# ==============================================================================
# Dataset N=1000 usuarios con CSAT y frecuencia de uso
N_csat = 1000
df_csat_pop = pd.DataFrame({
    'User_ID': range(1, N_csat + 1),
    'CSAT': np.clip(np.random.normal(loc=76, scale=15, size=N_csat), 1, 100).round(1),
    'Sesiones_Mes': np.random.poisson(lam=18, size=N_csat)
})

# Muestreo Aleatorio Simple n=150
n_csat = 150
df_csat_srs = df_csat_pop.sample(n=n_csat, replace=False, random_state=SEED)

print("--- EJEMPLO 2.1: CSAT APP MÓVIL ---")
print(f"Población CSAT Media: {df_csat_pop['CSAT'].mean():.2f} pts | Muestra SRS: {df_csat_srs['CSAT'].mean():.2f} pts")

fig, ax = plt.subplots(figsize=(10, 4))
sns.kdeplot(df_csat_pop['CSAT'], ax=ax, label='Población Completa (N=1000)', color=PRIMARY_COLOR, fill=True, alpha=0.3)
sns.kdeplot(df_csat_srs['CSAT'], ax=ax, label='Muestra SRS (n=150)', color=SAMPLE_COLOR, fill=True, alpha=0.4)
ax.set_title("2.1 Muestreo Aleatorio Simple: Densidad de Satisfacción (CSAT)")
ax.set_xlabel("CSAT Score (0 a 100)")
ax.legend()
plt.show()
````

````python
# ==============================================================================
# EJEMPLO 2.3: Auditoría de Transacciones Financieras y Detección de Irregularidades
# ==============================================================================
# N=1000 transacciones con 5% de anomalías
anomalias = np.random.choice([0, 1], size=N, p=[0.95, 0.05])
montos = np.where(anomalias == 1, np.random.uniform(5000, 15000, size=N), np.random.exponential(scale=350, size=N) + 10)

df_tx_pop = pd.DataFrame({
    'Tx_ID': [f"TX-{i:04d}" for i in range(1, N + 1)],
    'Monto_USD': montos,
    'Es_Anomalia': anomalias
})

# Muestra auditoría n=100
n_audit = 100
df_tx_sample = df_tx_pop.sample(n=n_audit, random_state=SEED)

p_real_anomalias = df_tx_pop['Es_Anomalia'].mean()
p_muestral_anomalias = df_tx_sample['Es_Anomalia'].mean()

print(f"--- EJEMPLO 2.3: AUDITORÍA FINANCIERA ---")
print(f"Tasa real de anomalías en población: {p_real_anomalias*100:.2f}%")
print(f"Tasa estimada en muestra SRS (n={n_audit}): {p_muestral_anomalias*100:.2f}%")

fig, ax = plt.subplots(figsize=(8, 4))
categorias = ['Población Real (N=1000)', 'Muestra SRS (n=100)']
proporciones = [p_real_anomalias * 100, p_muestral_anomalias * 100]
bars = ax.bar(categorias, proporciones, color=[PRIMARY_COLOR, SAMPLE_COLOR], width=0.5)
ax.set_ylabel("% Transacciones Anómalas")
ax.set_title("2.3 Estimación de Tasa de Fraude/Anomalía en Auditoría")
for bar in bars:
    yval = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2.0, yval + 0.2, f"{yval:.2f}%", ha='center', va='bottom', fontweight='bold')
plt.ylim(0, max(proporciones) + 3)
plt.show()
````

````python
# ==============================================================================
# EJEMPLO 2.4: Estudio de Convergencia del Error Muestral vs Tamaño de Muestra (n)
# ==============================================================================
# Demostración del Teorema Central del Límite: ¿Cómo evoluciona el error al variar n?
tamanos_n = np.arange(10, 500, 10)
errores = []

for n_test in tamanos_n:
    # Promedio de error absoluto en 50 réplicas por cada tamaño n
    errores_replica = [abs(df_csat_pop['CSAT'].sample(n=n_test).mean() - df_csat_pop['CSAT'].mean()) for _ in range(50)]
    errores.append(np.mean(errores_replica))

plt.figure(figsize=(10, 5))
plt.plot(tamanos_n, errores, marker='o', color=PRIMARY_COLOR, linewidth=2)
plt.title("2.4 Disminución del Error de Estimación al Incrementar el Tamaño de Muestra (n)")
plt.xlabel("Tamaño de Muestra (n)")
plt.ylabel("Error Absoluto Medio (|x̄ - μ|)")
plt.axvline(150, color='red', linestyle='--', label='Punto de Rendimiento Óptimo (n=150)')
plt.legend()
plt.show()
````

# ⏱️ 3. Muestreo Sistemático (Systematic Sampling)

### 📖 Concepto:
En el **Muestreo Sistemático**, los elementos se seleccionan a intervalos regulares a partir de un listado ordenado o un flujo continuo de datos. 

### ⚙️ Procedimiento:
1. Se determina el tamaño poblacional $N$ y el tamaño de muestra deseado $n$.
2. Se calcula el **intervalo de salto**: $k = \lfloor \frac{N}{n} \rfloor$.
3. Se selecciona aleatoriamente un punto de arranque $r$ (donde $1 \le r \le k$).
4. La muestra estará conformada por los elementos en las posiciones:
   $$\text{Muestra} = \{r, \, r + k, \, r + 2k, \, r + 3k, \dots, \, r + (n-1)k\}$$

### ⚠️ El Peligro Crítico: La Periodicidad Oculta
* Si los datos presentan un ciclo o estacionalidad oculta que coincide con el intervalo $k$ (o un múltiplo/divisor), la muestra estará **severamente sesgada**.
* *Ejemplo*: Si se inspecciona una máquina cada 8 horas y la máquina experimenta una subida de tensión cada 8 horas, la muestra solo registrará los picos de falla o solo los momentos óptimos.

Veamos **4 ejemplos aplicados** con $N = 1,000$ observaciones.



````python
# ==============================================================================
# FUNCIÓN UTILITARIA: Muestreo Sistemático
# ==============================================================================
def systematic_sampling(df, n, random_state=SEED):
    N = len(df)
    k = N // n
    np.random.seed(random_state)
    start = np.random.randint(0, k)
    indices = np.arange(start, N, k)[:n]
    return df.iloc[indices].copy(), k, start

print("✅ Función de Muestreo Sistemático definida.")
````

````python
# ==============================================================================
# EJEMPLO 3.1: Control de Calidad en Línea de Producción Industrial Continua
# ==============================================================================
# N=1000 piezas producidas en una jornada con diámetro nominal de 50 mm
diametros = np.random.normal(loc=50.0, scale=0.15, size=N)
df_produccion = pd.DataFrame({
    'Orden_Produccion': range(1, N + 1),
    'Diametro_mm': diametros
})

n_prod = 100 # Queremos inspeccionar 100 piezas (k = 1000/100 = 10)
df_muestra_sist, k_prod, start_prod = systematic_sampling(df_produccion, n=n_prod, random_state=SEED)

mu_diam = df_produccion['Diametro_mm'].mean()
xbar_diam = df_muestra_sist['Diametro_mm'].mean()

print(f"--- EJEMPLO 3.1: PRODUCCIÓN EN LÍNEA ---")
print(f"Intervalo k = {k_prod} | Punto de inicio aleatorio = {start_prod}")
print(f"Media Real μ: {mu_diam:.4f} mm | Media Sistemática x̄: {xbar_diam:.4f} mm")

plt.figure(figsize=(14, 4))
plt.plot(df_produccion['Orden_Produccion'], df_produccion['Diametro_mm'], color='lightgray', label='Flujo continuo de piezas (N=1000)')
plt.scatter(df_muestra_sist['Orden_Produccion'], df_muestra_sist['Diametro_mm'], color=SAMPLE_COLOR, s=25, zorder=5, label=f'Muestra Sistemática (cada k={k_prod} piezas)')
plt.title("3.1 Muestreo Sistemático en Línea de Manufactura")
plt.xlabel("Pieza Producida (Secuencia Temporal)")
plt.ylabel("Diámetro (mm)")
plt.legend()
plt.show()
````

# 📑 4. Muestreo Estratificado (Stratified Sampling)

### 📖 Concepto:
El **Muestreo Estratificado** se aplica cuando la población se compone de subgrupos o capas mutuamente excluyentes llamados **estratos** que presentan:
* **Alta homogeneidad interna**: Los individuos dentro de un mismo estrato se parecen mucho entre sí.
* **Alta heterogeneidad entre estratos**: Los estratos son marcadamente diferentes unos de otros.

### ⚖️ Asignación de la Muestra:
1. **Asignación Proporcional**: El tamaño de muestra $n_h$ en cada estrato $h$ es directamente proporcional al peso del estrato en la población ($W_h = N_h / N$):
   $$n_h = n \times \frac{N_h}{N}$$
2. **Asignación Óptima (Neyman)**: Considera no solo el tamaño del estrato sino también su variabilidad interna (desviación estándar $\sigma_h$):
   $$n_h = n \times \frac{N_h \sigma_h}{\sum N_i \sigma_i}$$

### 🎯 Diferencia Crucial con MAS:
En Muestreo Aleatorio Simple, un estrato minoritario (ej. 5% de clientes corporativos) corre el riesgo de no salir en la muestra o quedar subrepresentado. El muestreo estratificado **garantiza representación exacta** y **reduce drásticamente la varianza del estimador**.

Veamos **4 ejemplos aplicados** con $N = 1,000$ observaciones.



````python
# ==============================================================================
# FUNCIÓN UTILITARIA: Muestreo Estratificado Proporcional
# ==============================================================================
def stratified_sampling(df, group_col, n_total, random_state=SEED):
    # Proporciones por grupo
    proporciones = df[group_col].value_counts(normalize=True)
    muestras = []
    for grupo, prop in proporciones.items():
        n_grupo = int(round(prop * n_total))
        df_grupo = df[df[group_col] == grupo]
        muestras.append(df_grupo.sample(n=min(n_grupo, len(df_grupo)), random_state=random_state))
    return pd.concat(muestras).reset_index(drop=True)

print("✅ Función de Muestreo Estratificado definida.")
````

````python
# ==============================================================================
# EJEMPLO 4.1: Churn y Gasto en SaaS por Tier de Suscripción (Free, Pro, Enterprise)
# ==============================================================================
# N=1000 usuarios con proporciones: Free (60%), Pro (30%), Enterprise (10%)
n_free, n_pro, n_ent = 600, 300, 100
df_free = pd.DataFrame({'Plan': 'Free', 'Gasto_USD': np.random.normal(0, 0, n_free), 'Nivel_Uso': np.random.normal(20, 5, n_free)})
df_pro = pd.DataFrame({'Plan': 'Pro', 'Gasto_USD': np.random.normal(49, 5, n_pro), 'Nivel_Uso': np.random.normal(65, 10, n_pro)})
df_ent = pd.DataFrame({'Plan': 'Enterprise', 'Gasto_USD': np.random.normal(499, 40, n_ent), 'Nivel_Uso': np.random.normal(95, 8, n_ent)})

df_saas_pop = pd.concat([df_free, df_pro, df_ent]).sample(frac=1, random_state=SEED).reset_index(drop=True)

# Muestra Estratificada vs Muestra Simple (n=100)
n_saas = 100
df_saas_strat = stratified_sampling(df_saas_pop, 'Plan', n_total=n_saas, random_state=SEED)
df_saas_srs = df_saas_pop.sample(n=n_saas, random_state=SEED)

print("--- EJEMPLO 4.1: SAAS TIERS ---")
print("Distribución en Población:  
", df_saas_pop['Plan'].value_counts(normalize=True))
print("Distribución en Estratificado:
", df_saas_strat['Plan'].value_counts(normalize=True))
print("Distribución en Aleatorio Simple:
", df_saas_srs['Plan'].value_counts(normalize=True))

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Comparación de proporciones
df_comp_prop = pd.DataFrame({
    'Población (N=1000)': df_saas_pop['Plan'].value_counts(normalize=True)*100,
    'Estratificado (n=100)': df_saas_strat['Plan'].value_counts(normalize=True)*100,
    'Aleatorio Simple (n=100)': df_saas_srs['Plan'].value_counts(normalize=True)*100
})
df_comp_prop.plot(kind='bar', ax=ax1, color=[PRIMARY_COLOR, ACCENT_COLOR, SAMPLE_COLOR])
ax1.set_title("4.1 Representatividad Exacta por Tier")
ax1.set_ylabel("% del Total")
ax1.set_xticklabels(ax1.get_xticklabels(), rotation=0)

# Violinplot por estrato
sns.violinplot(data=df_saas_pop, x='Plan', y='Gasto_USD', ax=ax2, palette='Blues')
ax2.set_title("Heterogeneidad Marcada Entre Estratos de Planes")
plt.tight_layout()
plt.show()
````

# 🏢 5. Muestreo por Conglomerados (Cluster Sampling)

### 📖 Concepto:
El **Muestreo por Conglomerados** se utiliza cuando la población está dividida de forma natural en grupos o clusters geográficos, organizacionales o temporales. 

A diferencia del muestreo estratificado:
* **Alta heterogeneidad intra-conglomerado**: Cada conglomerado es un "microcosmos" o reflejo representativo de toda la población.
* **Alta homogeneidad inter-conglomerados**: Todos los conglomerados son muy parecidos entre sí en su estructura global.

### 🔍 Diferencia Clave entre Estratificado y Conglomerados:

| Criterio | Muestreo Estratificado | Muestreo por Conglomerados |
| :--- | :--- | :--- |
| **Composición interna** | Elementos homogéneos dentro del estrato | Elementos heterogéneos dentro del cluster (un mini-universo) |
| **Diferencia entre grupos** | Estratos muy distintos entre sí | Conglomerados muy similares entre sí |
| **Selección** | Se toma una muestra de **TODOS** los estratos | Se eligen **ALGUNOS** conglomerados al azar y los demás se descartan |
| **Objetivo principal** | Maximizar precisión y representatividad | Minimizar costos operativos y logísticos de traslado/recolección |

### 🪜 Tipos de Muestreo por Conglomerados:
1. **Una Etapa (Single-Stage)**: Se seleccionan $m$ conglomerados al azar y se analizan **todos** los individuos dentro de ellos.
2. **Dos Etapas (Two-Stage)**: Se seleccionan $m$ conglomerados al azar y, dentro de cada uno, se realiza un muestreo aleatorio simple para seleccionar solo una fracción de individuos.

Veamos **4 ejemplos aplicados** con $N = 1,000$ observaciones agrupadas.



````python
# ==============================================================================
# EJEMPLO 5.1: Auditoría Presencial en Sucursales de Tiendas Departamentales (Una Etapa)
# ==============================================================================
# N=1000 empleados repartidos en 50 sucursales (20 empleados por sucursal)
num_sucursales = 50
empleados_por_sucursal = 20

sucursal_ids = np.repeat([f"Sucursal_{i:02d}" for i in range(1, num_sucursales + 1)], empleados_por_sucursal)
satisfaccion_empleados = np.random.normal(loc=75, scale=12, size=N)

df_retail_pop = pd.DataFrame({
    'Sucursal': sucursal_ids,
    'Empleado_ID': range(1, N + 1),
    'Satisfaccion': np.clip(satisfaccion_empleados, 1, 100)
})

# Muestreo por Conglomerados en 1 Etapa: Elegimos 6 sucursales al azar de las 50 (n = 6 * 20 = 120 empleados)
np.random.seed(SEED)
sucursales_elegidas = np.random.choice(df_retail_pop['Sucursal'].unique(), size=6, replace=False)
df_cluster_stage1 = df_retail_pop[df_retail_pop['Sucursal'].isin(sucursales_elegidas)].copy()

print(f"--- EJEMPLO 5.1: SUCURSALES RETAIL (1 ETAPA) ---")
print(f"Sucursales seleccionadas ({len(sucursales_elegidas)} de {num_sucursales}): {list(sucursales_elegidas)}")
print(f"Total empleados evaluados: {len(df_cluster_stage1)} de {N}")
print(f"Media Poblacional: {df_retail_pop['Satisfaccion'].mean():.2f} | Media Conglomerado: {df_cluster_stage1['Satisfaccion'].mean():.2f}")

plt.figure(figsize=(14, 5))
todos_los_clusters = df_retail_pop['Sucursal'].unique()[:20] # Mostramos primeras 20 para claridad
colores = [CLUSTER_COLOR if s in sucursales_elegidas else 'lightgray' for s in todos_los_clusters]
df_subset = df_retail_pop[df_retail_pop['Sucursal'].isin(todos_los_clusters)]

sns.boxplot(data=df_subset, x='Sucursal', y='Satisfaccion', palette=colores)
plt.title("5.1 Muestreo por Conglomerados: Sucursales Seleccionadas (Rojo) vs No Visitadas (Gris)")
plt.xticks(rotation=45)
plt.show()
````

````python
# ==============================================================================
# EJEMPLO 5.2: Encuesta de Desempeño a Alumnos en Aulas de Colegios
# ==============================================================================
# N=1000 alumnos en 40 aulas (25 alumnos por aula)
num_aulas = 40
alumnos_por_aula = 25

aulas = np.repeat([f"Aula_{i:02d}" for i in range(1, num_aulas + 1)], alumnos_por_aula)
calificaciones = np.random.normal(loc=82, scale=10, size=N)

df_escuelas_pop = pd.DataFrame({'Aula': aulas, 'Nota': calificaciones})

# Seleccionamos 5 aulas completas (n = 5 * 25 = 125 alumnos)
np.random.seed(SEED)
aulas_seleccionadas = np.random.choice(df_escuelas_pop['Aula'].unique(), size=5, replace=False)
df_aulas_cluster = df_escuelas_pop[df_escuelas_pop['Aula'].isin(aulas_seleccionadas)]

mu_notas = df_escuelas_pop['Nota'].mean()
xbar_cluster_notas = df_aulas_cluster['Nota'].mean()

fig, ax = plt.subplots(figsize=(10, 4))
sns.kdeplot(df_escuelas_pop['Nota'], color=PRIMARY_COLOR, fill=True, alpha=0.3, label=f'Población (40 aulas, μ={mu_notas:.2f})')
sns.kdeplot(df_aulas_cluster['Nota'], color=CLUSTER_COLOR, fill=True, alpha=0.4, label=f'Conglomerado (5 aulas, x̄={xbar_cluster_notas:.2f})')
ax.set_title("5.2 Comparación de Distribución de Notas: Población vs Conglomerados de Aulas")
ax.set_xlabel("Calificación")
ax.legend()
plt.show()
````

````python
# ==============================================================================
# EJEMPLO 5.4: Muestreo en Dos Etapas (Two-Stage Cluster) en Rutas de Entrega Urbana
# ==============================================================================
# N=1000 paquetes repartidos en 20 zonas urbanas (50 paquetes por zona)
num_zonas = 20
pkgs_por_zona = 50

zonas = np.repeat([f"Zona_{i:02d}" for i in range(1, num_zonas + 1)], pkgs_por_zona)
tiempos_entrega = np.random.normal(loc=35, scale=8, size=N)
df_logistica_pop = pd.DataFrame({'Zona': zonas, 'Tiempo_Min': tiempos_entrega})

# ETAPA 1: Seleccionar 4 zonas al azar
np.random.seed(SEED)
zonas_elegidas = np.random.choice(df_logistica_pop['Zona'].unique(), size=4, replace=False)

# ETAPA 2: Dentro de cada una de las 4 zonas elegidas, seleccionar 25 paquetes al azar (Total n = 4 * 25 = 100)
muestras_etapa2 = []
for z in zonas_elegidas:
    df_zona = df_logistica_pop[df_logistica_pop['Zona'] == z]
    muestras_etapa2.append(df_zona.sample(n=25, random_state=SEED))

df_two_stage = pd.concat(muestras_etapa2).reset_index(drop=True)

print(f"--- EJEMPLO 5.4: MUESTREO EN DOS ETAPAS ---")
print(f"Zonas seleccionadas (Etapa 1): {list(zonas_elegidas)}")
print(f"Total paquetes evaluados (Etapa 2): {len(df_two_stage)} de {N}")
print(f"Media Real: {df_logistica_pop['Tiempo_Min'].mean():.2f} min | Media Two-Stage: {df_two_stage['Tiempo_Min'].mean():.2f} min")

fig, ax = plt.subplots(figsize=(10, 4))
sns.boxplot(data=df_two_stage, x='Zona', y='Tiempo_Min', palette='Reds', ax=ax)
ax.set_title("5.4 Muestreo en Dos Etapas: Submuestras Seleccionadas por Zona")
ax.set_ylabel("Tiempo de Entrega (Minutos)")
plt.show()
````

# 🔬 6. Simulación Monte Carlo y Comparativa Final

### 🧪 El Experimento:
Para evaluar la precisión, varianza y error de cada técnica de muestreo, someteremos una población heterogénea ($N=1000$) a **1,000 repeticiones de muestreo** con un tamaño de muestra fijo $n=100$.

Calcularemos:
* **Sesgo (Bias)**: $E[\bar{x}] - \mu$ (debe ser cercano a 0).
* **Varianza del Estimador**: $Var(\bar{x})$.
* **Error Cuadrático Medio (MSE)**: $MSE = \text{Var} + \text{Sesgo}^2$.



````python
# ==============================================================================
# SIMULACIÓN MONTE CARLO: 1000 Iteraciones
# ==============================================================================
# Creamos una población realista con 4 estratos y 20 conglomerados
np.random.seed(SEED)
N_sim = 1000
estratos_sim = np.random.choice(['A', 'B', 'C', 'D'], size=N_sim, p=[0.4, 0.3, 0.2, 0.1])
conglomerados_sim = np.repeat(range(1, 21), 50) # 20 clusters de 50 elementos

valores_sim = []
for e in estratos_sim:
    if e == 'A': valores_sim.append(np.random.normal(50, 10))
    elif e == 'B': valores_sim.append(np.random.normal(70, 12))
    elif e == 'C': valores_sim.append(np.random.normal(100, 15))
    else: valores_sim.append(np.random.normal(140, 20))

df_benchmark_pop = pd.DataFrame({
    'Valor': valores_sim,
    'Estrato': estratos_sim,
    'Cluster': conglomerados_sim
})

mu_benchmark = df_benchmark_pop['Valor'].mean()
n_sim = 100
num_iteraciones = 1000

medias_srs = []
medias_sist = []
medias_strat = []
medias_cluster = []

for i in range(num_iteraciones):
    # 1. Aleatorio Simple
    srs = df_benchmark_pop.sample(n=n_sim, random_state=i)
    medias_srs.append(srs['Valor'].mean())
    
    # 2. Sistemático
    sist, _, _ = systematic_sampling(df_benchmark_pop, n=n_sim, random_state=i)
    medias_sist.append(sist['Valor'].mean())
    
    # 3. Estratificado
    strat = stratified_sampling(df_benchmark_pop, 'Estrato', n_total=n_sim, random_state=i)
    medias_strat.append(strat['Valor'].mean())
    
    # 4. Conglomerados (Elegir 2 clusters de 50 = 100)
    c_chosen = np.random.choice(range(1, 21), size=2, replace=False)
    cluster = df_benchmark_pop[df_benchmark_pop['Cluster'].isin(c_chosen)]
    medias_cluster.append(cluster['Valor'].mean())

# Métricas de Desempeño
resultados = pd.DataFrame({
    'Método': ['Aleatorio Simple (MAS)', 'Sistemático', 'Estratificado', 'Conglomerados'],
    'Media de Estimaciones': [np.mean(medias_srs), np.mean(medias_sist), np.mean(medias_strat), np.mean(medias_cluster)],
    'Varianza Muestral': [np.var(medias_srs), np.var(medias_sist), np.var(medias_strat), np.var(medias_cluster)],
    'Desv. Estándar (SE)': [np.std(medias_srs), np.std(medias_sist), np.std(medias_strat), np.std(medias_cluster)],
    'MSE': [np.mean((np.array(m) - mu_benchmark)**2) for m in [medias_srs, medias_sist, medias_strat, medias_cluster]]
})

print(f"🎯 MEDIA POBLACIONAL REAL μ = {mu_benchmark:.4f}
")
print(resultados.to_string(index=False))

# Visualización de Distribuciones Muestrales de las Medias
plt.figure(figsize=(14, 6))
sns.kdeplot(medias_strat, label=f'Estratificado (Varianza mínima: {np.var(medias_strat):.3f})', color=ACCENT_COLOR, fill=True, alpha=0.3, linewidth=2)
sns.kdeplot(medias_srs, label=f'Aleatorio Simple (Varianza: {np.var(medias_srs):.3f})', color=PRIMARY_COLOR, fill=True, alpha=0.2, linewidth=2)
sns.kdeplot(medias_sist, label=f'Sistemático (Varianza: {np.var(medias_sist):.3f})', color=SAMPLE_COLOR, fill=True, alpha=0.2, linewidth=2)
sns.kdeplot(medias_cluster, label=f'Conglomerados (Mayor dispersión: {np.var(medias_cluster):.3f})', color=CLUSTER_COLOR, fill=True, alpha=0.1, linewidth=2)
plt.axvline(mu_benchmark, color='black', linestyle='--', linewidth=2, label=f'Media Poblacional Real μ={mu_benchmark:.2f}')

plt.title("6. Distribución Muestral de las Medias en 1,000 Simulaciones Monte Carlo")
plt.xlabel("Valor de la Media Estimada (x̄)")
plt.ylabel("Densidad de Probabilidad")
plt.legend()
plt.show()
````

# 📋 7. Guía de Decisión y Conclusiones para Ciencia de Datos

### 🧭 ¿Qué método de muestreo elegir en tu proyecto?

```mermaid
graph TD
    A[¿Tienes acceso al marco muestral completo?] -->|No| B[Muestreo No Probabilístico / Oportunidad]
    A -->|Sí| C{¿La población tiene subgrupos conocidos y heterogéneos?}
    C -->|Sí| D[Muestreo Estratificado: Mínima Varianza y Máxima Representatividad]
    C -->|No| E{¿La recolección es en flujo continuo o proceso secuencial?}
    E -->|Sí| F[Muestreo Sistemático: Cuidado con la periodicidad]
    E -->|No| G{¿Hay restricciones geográficas o de costos logísticos?}
    G -->|Sí| H[Muestreo por Conglomerados: Menor costo, mayor varianza]
    G -->|No| I[Muestreo Aleatorio Simple: Insesgado y directo]
```

### 💡 Resumen Ejecutivo:
1. **Población vs. Muestra**: Los parámetros poblacionales son fijos; los estadísticos muestrales son variables aleatorias sujetas a error estándar que disminuye a tasa $1/\sqrt{n}$.
2. **Aleatorio Simple**: Excelente estándar general, pero ineficiente cuando existen minorías críticas o agrupaciones geográficas.
3. **Sistemático**: Ideal para pipelines de datos en streaming y líneas de producción; siempre verificar ausencia de periodicidad.
4. **Estratificado**: La mejor opción para maximizar precisión cuando los grupos son heterogéneos entre sí.
5. **Conglomerados**: La mejor opción para optimizar costos de recolección en campo cuando los grupos son mini-poblaciones similares entre sí.
